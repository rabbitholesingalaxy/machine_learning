```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# the following are multinomial naive bayes tutorial
data={
    'text':[
        'Free money now',
        'Call now to claim your prize',
        'Meet me at the park',
        'Let’s catch up later',
        'Win a new car today!',
        'Lunch plans?',
        'Congratulations! You won a lottery',
        'Can you send me the report?',
        'Exclusive offer for you',
        'Are you coming to the meeting?'
    ],
    'label':[
        'spam', 'spam', 'not spam', 'not spam', 'spam', 'not spam', 'spam', 'not spam', 'spam', 'not spam'
    ]
}
df=pd.DataFrame(data)
```

```python
df['label']=df['label'].map({'spam':1,'not spam':0})
x=df['text']
y=df['label']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
vectorizer=CountVectorizer()
# Behavior: Treats all frequent words as important. Words that appear in almost every document (like "the" or "is") get high counts and can drown out rare, meaningful words.
# text->vector by counting the occurrence of each word
x_train_vectors=vectorizer.fit_transform(x_train)
x_test_vectors=vectorizer.transform(x_test)

model=MultinomialNB()
model.fit(x_train_vectors,y_train)
# use multinomial train the model 

y_pred=model.predict(x_test_vectors)
accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy: {accuracy*100:.2f}%\n")

```

```python
custom_message=["Congratulations, you've won a free vacation"]
custom_vector=vectorizer.transform(custom_message)
prediction=model.predict(custom_vector)
print("the message is", "spam" if prediction[0]==1 else "Not spam")
```
