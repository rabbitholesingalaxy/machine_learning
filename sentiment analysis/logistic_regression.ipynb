```python
# log-odds: logit(p)= ln(p/(1-p))
# odds: p/(1-p) success/failure
# transform bounded probability(0,1) to unbounded log-odds(-inf, inf), which can be modeled as a linear function of the predictors.

# result interpretation:
# 0 means p=0.5, 1 means p>0.5, -1 means p<0.5
```

```python
from sklearn.datasets import load_breast_cancer

#  This function returns a dictionary-like object containing the features and target labels of the dataset. You can also specify whether to return the data as a pandas DataFrame or as NumPy arrays.
X,y = load_breast_cancer(return_X_y=True, as_frame=True)
y.tail()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)

from sklearn.linear_model import LogisticRegression
binomial_model = LogisticRegression(max_iter=1000, random_state=42)
binomial_model.fit(X_train, y_train)

from sklearn.metrics import classification_report
y_pred_bi= binomial_model.predict(X_test)
print(classification_report(y_test, y_pred_bi))
```
