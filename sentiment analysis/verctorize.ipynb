```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# the following are verctorize tutorial
from collections import Counter
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
]
all_words=''.join(corpus).lower().split()
counts=Counter(all_words)
# does not raise an error if dictionary not found, also can add/deduct
print(counts)
print(counts.most_common(2))
print(counts.most_common()[-1])
print(counts.most_common()[:-3:-1])
```

```python
cv=CountVectorizer()
# stop_words='english': Automatically removes common, low-information English words (like "the", "and", "is") so they don't clutter your data.
# generate a dictionary of the frequency of a word, where word is vectorized to vector 
X_counts=cv.fit_transform(corpus)
# Fit the model and transform the text into counts
df_counts=pd.DataFrame(X_counts.toarray(),columns=cv.get_feature_names_out())
# columns are the unqiue word appear in the document, rows are document
print(df_counts)
print(df_counts.sum(axis=0).sort_values(ascending=False))
```

```python
vectorizer=TfidfVectorizer()
X=vectorizer.fit_transform(corpus)
test=pd.DataFrame(X.toarray(),columns=vectorizer.get_feature_names_out())
print(test)
# rows: document/ columns: vocabulary terms/cell value:TF-IDF score 
# importance ranking
# High (~0.5)	Term is distinctive to this document
# Medium (~0.3)	Term appears in this doc and a few others
# Zero	Term not present in this document
```

```python
for i,row in test.iterrows():
 #  test.iterrows() yields a tuple containing exactly two elements: (index, row_series)
    print(f"---Document{i}---")
    print(row.sort_values(ascending=False))
    print()
```

```python
top_n=3
for i, row in test.iterrows():
    top = row.sort_values(ascending=False).head(top_n)
    print(f"Document {i}: {list(top.index)}")
# index-> column name, value-> data
```

```python
for i,row in test.iterrows():
# i: index row:column name:value
    nonzero=row[row>0].sort_values(ascending=False)
    print(f"---Document{i}---")
    print(nonzero)
```

```python
print(test.T.sort_values(by=0,ascending=False))
# T->transpose
# by=0, sorting by column 0 descending way
print(test[['first','second','third']])
```
