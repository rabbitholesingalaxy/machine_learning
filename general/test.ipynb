```python
# Unparametric test-u test 
a=[85, 88, 90, 92, 95]
b=[78, 80, 83, 85, 87]
import scipy.stats as stats
stat,p_value=stats.mannwhitneyu(a,b,alternative='two-sided')
# alternative='two-sided'->whether are different --> accurate, default 
# alternative='greater'->whether a greater than b
# alternative='less'->whether a less than b 
print(f"U statistic:{stat:.2f},p-value{p_value:.2f}")
# p<0.05, reject the null hypothesis--> significant difference between two regions 
alpha=0.05
if p_value<alpha:
    print('reject null hypothesis')
else:
    print('support null hypothesis')
```

```python
# parametric t test: require normally distributed 
# 1. independent sample
import numpy as np
a=np.array([88, 92, 78, 85, 90, 84, 87, 94, 80, 89])
b=np.array([82, 85, 75, 81, 79, 86, 83, 88, 74, 80])
t_stat,p_value=stats.ttest_ind(a,b,equal_var=False,alternative='less')
# equal variance=True--> student T-Test, equal variance=False--> Welch's T-Test
# set False, welch works in both, student fail in unequal variance 
print(f"T statistic:{t_stat:.2f},p-value:{p_value:.2f}")
# 2. paired sample 
before=np.array([120, 122, 115, 124, 130, 118, 121, 128])
after=np.array([115, 118, 112, 120, 122, 116, 115, 124])
t_stat,p_value=stats.ttest_rel(before,after)
print(f"T statistic:{t_stat:.2f},p-value:{p_value:.4f}")
# 3. one sample
scores=np.array([75, 78, 82, 79, 85, 88, 76, 80, 81, 79])
known_mean=75
t_stat,p_value=stats.ttest_1samp(scores,popmean=known_mean)
print(f"T statistic:{t_stat:.2f},p-value:{p_value:.4f}")
```
