```python
# Single split	                                            Cross-validation
# One number, depends on which rows landed where	        Average of k numbers → less luck-dependent
# No sense of variability	                                a standard deviation
# Wastes data (20% locked away)	                            Every row is used for validation exactly once
# fast and easy                                             slower and more complex
# to generate a confusion matrix or visualize specific classification errors
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score

iris=load_iris()

from pandas import DataFrame
df=DataFrame(iris.data,columns=iris.feature_names)
df['target']=iris.target
df.tail()
```

```python
from sklearn.svm import SVC
svm_classifier=SVC(kernel='linear',C=1)
```

```python
# cross-validation
num_fold=5
# Few Folds (High Bias):When k is small (e.g., k=2), each training fold uses only a fraction of the total data (e.g., 50%).
# Too Many Folds (High Variance):When k is very large, each of the k training sets shares almost identical data points, differing by only one or a few samples.
# split the dataset into 5 folds, train on 4 folds and test on the remaining fold, repeat this process 5 times so that each fold is used as a test set once.
from sklearn.model_selection import KFold
kf=KFold(n_splits=num_fold,shuffle=True,random_state=42)
cv_score=cross_val_score(svm_classifier,iris.data,iris.target,cv=kf)
# cv:cross validation =kv : parameter define n-split/shuffle and random_state
print("cross-validation results")
for i, result in enumerate(cv_score,1):
# 1: start number
# i->1, result->score[0]; i->2,result->score[1]; i->3,result->score[2]; i->4,result->score[3]; i->5,result->score[4]
# a cleaner and efficient way than for i in range(len(score)):
# enumerate(score,1) will give you the index (starting from 1) and the corresponding result from the score array. The loop will iterate through each fold's score and print it in a formatted manner.
    print(f"Fold {i}: {result*100:.2f}%")
print(f"Mean cross-validation score: {cv_score.mean()*100:.2f}%")
# mean cross-validation score represents the model's overall expected performance on unseen data, 
# while the standard deviation indicates how much the performance varies across different folds. A lower standard deviation suggests that the model's performance is more consistent across different subsets of the data, while a higher standard deviation indicates that the model's performance is more sensitive to the specific data used for training and testing.
print(f"Standard deviation: {cv_score.std()*100:.2f}%")
# if the results prove model is reliable, then we can use the model (with the same parameters) to predict on new unseen data.
# Once you are satisfied with the cross-validation scores and are ready to deploy or make predictions, you must explicitly fit your model on the entire dataset
# cross_val_score only returns an array of scores to help you evaluate performance, it does not return a trained model.
print(f"mean accuracy: {cv_score.mean()*100:.2f}%(std: {cv_score.std()*100:.2f}%)")
# the cross_val_score packed containing:
# Common Classification Scores
# • Accuracy: The percentage of correct predictions out of all predictions.
# • F1-Score / Macro F1: A balanced measure of precision and recall, ideal for imbalanced classes.
# • Precision & Recall: Measures of exactness (false positives) and completeness (false negatives).
# • ROC-AUC: Measures a model's ability to distinguish between classes.
# Common Regression Scores
# • Mean Squared Error (MSE) / Root Mean Squared Error (RMSE): Measures the average magnitude of the errors.
# • Mean Absolute Error (MAE): The average absolute distance between predicted and actual values.
# • R² (Coefficient of Determination): Represents the proportion of variance in the dependent variable explained by the model.
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,:2], df['target'], test_size=0.2, random_state=42)
svm_classifier.fit(X_train, y_train)
test_score=svm_classifier.score(X_test, y_test)
print(f"Mean train_test_split score: {test_score*100:.2f}%")
```
